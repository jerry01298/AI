```
from tensorflow import keras
from keras.models import Sequential
from keras.layers import Dense

import numpy as np

data = np.random.random((1000, 10))  #隨機產生1000筆資料，每筆有10個特徵
labels = np.random.randint(2, size=(1000, 1))  #隨機產生1000筆標籤，數值為0或1

print(np.shape(data))  #印出資料的形狀
print(data[0])      #印出第一筆資料
print(np.shape(labels)) #印出標籤的形狀
print(labels[0])     #印出第一筆標籤

model = Sequential()  #建立循序神經網路模型

model.add(Dense(32, activation='relu', input_dim=10))  #建立隱藏層，包含32個神經元，輸入10個特徵，使用ReLU激活函數

model.add(Dense(1, activation='sigmoid'))  #建立輸出層，使用1個神經元，透過Sigmoid輸出0到1之間的數值

model.compile(optimizer='rmsprop', loss='binary_crossentropy', metrics=['accuracy'])  #設定優化器、損失函數及準確率評估方式

model.fit(data, labels, epochs=10, batch_size=32)  #開始訓練模型，共訓練10次，每次使用32筆資料

score = model.evaluate(data, labels)  #使用資料評估模型的表現

print(score[1])  #印出模型的準確率
```
